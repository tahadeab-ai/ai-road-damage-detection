### 1. Negative Sample Integration
Add road imagery containing manholes, paint lines, and shadows labeled with 0 annotations.

```python
import albumentations as A
import cv2

# Advanced road ADAS augmentation pipeline
transform = A.Compose([
    A.RandomBrightnessContrast(p=0.4),
    A.MotionBlur(blur_limit=5, p=0.2),
    A.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, p=0.3),
    A.HorizontalFlip(p=0.5),
], bbox_params=A.BboxParams(format='yolo', label_fields=['class_labels']))
print('Augmentation pipeline initialized.')
```
