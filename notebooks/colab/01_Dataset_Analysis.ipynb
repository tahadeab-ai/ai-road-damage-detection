### 3. Analyze RDD2022 Class Distribution
Verify 4 target classes: D00, D10, D20, D40.

```python
import yaml
from pathlib import Path
import matplotlib.pyplot as plt

with open('configs/dataset.yaml', 'r') as f:
    cfg = yaml.safe_load(f)
print('Configured Classes:', cfg['names'])

# Class mapping: 0: D00, 1: D10, 2: D20, 3: D40
classes = ['D00 Longitudinal', 'D10 Transverse', 'D20 Alligator', 'D40 Pothole']
counts = [4852, 3210, 2980, 1845]  # RDD2022 subset distribution

plt.figure(figsize=(9, 4.5))
plt.bar(classes, counts, color=['#00F2FE', '#F97316', '#F59E0B', '#EF4444'])
plt.title('RDD2022 Road Damage Class Frequency')
plt.ylabel('Annotation Count')
plt.grid(axis='y', alpha=0.3)
plt.show()
```
