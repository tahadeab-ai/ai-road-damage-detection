## 02: Data Cleaning, Leakage Checks & Sequence Splitting

Validates image integrity, eliminates corrupted or zero-byte files, and generates sequence-based train/val splits to prevent data leakage.


### 1. Image Integrity and Dimension Checks

```python
from PIL import Image
from pathlib import Path

def verify_images(image_dir):
    corrupted = []
    for p in Path(image_dir).glob('*.jpg'):
        try:
            with Image.open(p) as img:
                img.verify()
        except Exception as e:
            corrupted.append((p, str(e)))
    print(f'Verified images. Corrupted found: {len(corrupted)}')
    return corrupted
```
