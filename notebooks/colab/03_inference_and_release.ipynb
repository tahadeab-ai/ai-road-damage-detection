# AI Road Damage Detection System
## Phase 3: Inference, Schema Serialization & GitHub Release Delivery

Covers:
1. Single image, batch directory, and streaming inference producing standardized JSON outputs.
2. Visualizer overlays with color-coded damage categories.
3. ONNX export with dynamic batch axes.
4. Automated GitHub Releases artifact publishing.

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd() / "src"))

from road_damage.inference.predictor import RoadDamagePredictor
from road_damage.utils.visualizer import RoadDamageVisualizer
from road_damage.models.detector import RoadDamageDetector
from road_damage.utils.github_release import GitHubReleasePublisher

# 1. Initialize Predictor with best weights
weights_path = "runs/detect/exp_a_nano_baseline/weights/best.pt"
predictor = RoadDamagePredictor(weights_or_model=weights_path if Path(weights_path).exists() else "yolo11n.pt")
```

```python
# 2. ONNX Export
detector = RoadDamageDetector(model_name_or_path=weights_path if Path(weights_path).exists() else "yolo11n.pt")
onnx_model_path = detector.export_onnx(imgsz=640)
print(f"Exported ONNX model: {onnx_model_path}")
```

```python
# 3. Automated GitHub Release Artifact Delivery
publisher = GitHubReleasePublisher(tag_name="v1.0.0-weights")
status = publisher.publish_bundle(
    best_pt=weights_path if Path(weights_path).exists() else None,
    model_onnx=onnx_model_path,
    confusion_matrix="runs/detect/exp_a_nano_baseline/confusion_matrix.png",
    results_curves="runs/detect/exp_a_nano_baseline/results.png",
    evaluation_summary="runs/detect/exp_a_nano_baseline/evaluation_summary.json"
)
print(f"Release Status: {status}")
```
