## 04: YOLO11n Mobile Model Training on RDD2022

Executes complete fine-tuning of YOLO11n on RDD2022 with optimal hyperparameter settings for mobile inference.


### 1. Initialize YOLO11n with Pretrained Weights

```python
from ultralytics import YOLO

model = YOLO('yolo11n.pt')
print('Loaded baseline YOLO11n model.')
```

### 2. Execute Training Loop

```python
results = model.train(
    data='configs/dataset.yaml',
    epochs=50,
    imgsz=640,
    batch=32,
    device=0,
    optimizer='AdamW',
    lr0=0.001,
    lrf=0.01,
    patience=15,
    mosaic=0.8,
    save=True,
    project='runs/train',
    name='roadguard_yolo11n'
)
```
